# 00 — Dataset overview

Discover run folders, inspect scenario metadata, and check which files are present.

```python
from pathlib import Path
import pandas as pd

from robustedge.data import DatasetIndex

DATA_ROOT = Path('../data/raw/logs')  # change if needed
if not DATA_ROOT.exists():
    DATA_ROOT = Path('../tests/fixtures/minimal_dataset')

index = DatasetIndex.from_root(DATA_ROOT)
run_index = index.to_frame()
print(f'Discovered {len(run_index)} runs under {DATA_ROOT.resolve()}')
run_index.head()
```

```python
if not run_index.empty:
    display(run_index.groupby(['perturbation', 'attack_duration', 'attack_intensity'], dropna=False).size().reset_index(name='n_runs'))
```

```python
# Inspect one loaded run
runs = index.load_runs()
run = runs[0]
print(run.run_id)
print(run.scenario)
print(run.sysdig.head())
```
