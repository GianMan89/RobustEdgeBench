# 04 — Paper figures

Generate robustness curves and timeline plots for the ETFA paper.

```python
from pathlib import Path
import pandas as pd
from robustedge.plotting import plot_robustness_curve, plot_timeline

OUTPUT_DIR = Path('../outputs/notebook_pipeline')
FIG_DIR = OUTPUT_DIR / 'figures_manual'
FIG_DIR.mkdir(parents=True, exist_ok=True)
agg = pd.read_csv(OUTPUT_DIR / 'metrics_aggregated.csv')
scores = pd.read_csv(OUTPUT_DIR / 'scores_by_window.csv')

if 'event_recall_mean' in agg.columns:
    for family in agg['perturbation_family'].dropna().unique():
        plot_robustness_curve(agg, 'event_recall_mean', 'event_recall_std', family=family, output_path=FIG_DIR / f'er_{family}.png', ylabel='Event recall')

if not scores.empty and (scores['label'] == 1).any():
    run_id = scores.loc[scores['label'] == 1, 'run_id'].iloc[0]
    detector = scores['detector'].iloc[0]
    plot_timeline(scores, run_id, detector, output_path=FIG_DIR / 'timeline_example.png')

print(f'Figures written to {FIG_DIR}')
```
