# 04 — Paper figures

Generate robustness heatmaps and timeline examples.

```python
from pathlib import Path
import pandas as pd
from robustedge.plotting import plot_heatmap, plot_profile, plot_timeline
OUTPUT_DIR = Path('../outputs/notebook_pipeline')
FIG_DIR = OUTPUT_DIR/'figures_manual'; FIG_DIR.mkdir(parents=True, exist_ok=True)
agg = pd.read_csv(OUTPUT_DIR/'metrics_aggregated.csv')
scores = pd.read_csv(OUTPUT_DIR/'scores_by_window.csv')
phase4 = agg[agg.phase == 'phase4_perturbed_attacked']
if not phase4.empty and 'event_recall_mean' in phase4.columns:
    det = phase4.detector.iloc[0]
    plot_heatmap(phase4, 'event_recall_mean', det, output_path=FIG_DIR/'event_recall_heatmap.png')
    plot_profile(phase4, 'event_recall_mean', output_path=FIG_DIR/'event_recall_profiles.png')
if not scores.empty and (scores.label == 1).any():
    row = scores[scores.label == 1].iloc[0]
    plot_timeline(scores, row.run_id, row.detector, output_path=FIG_DIR/'timeline_example.png')
print(f'Figures written to {FIG_DIR}')
```
